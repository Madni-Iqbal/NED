##### python f-Strings – Advanced Formatting

```python
from datetime import datetime

products = [
    {"name": "Iphone", "price": 1300, "release_date": datetime(2023, 5, 15)},
    {"name": "Book", "price": 79, "release_date": datetime(2022, 8, 20)},
    {"name": "Ticket", "price": 500, "release_date": datetime(2021, 12, 1)},
    {"name": "Clothes", "price": 80, "release_date": datetime(2024, 8, 13)},
    {"name": "Bag", "price": 55, "release_date": datetime(2019, 9, 14)}


]


print(f"{'Product Name':<15} {'Price':>10} {'Release Date':>15}")
print("-" * 45)


for product in products:
    name = product["name"]
    price = f"${product['price']:.2f}"  
    release_date = product["release_date"].strftime("%b %d, %Y")  
    print(f"{name:<15} {price:>10} {release_date:>15}")
```

### how f-strings improve code readability and formatting precision.

1.f-Strings make code easier to read by putting variables and expressions in the string directly.

2.They allow formatting to be done inline, which saves time and effort.

3.Code becomes shorter, cleaner, and easier to maintain.

##### Data Types – Lists, Dictionaries, Tuples, and Sets

```python
#1. Lists

student_names = ["Madni", "Farukh", "Fawwad", "Saad", "Fahad"]
print("Initial List:", student_names)


student_names.append("Ali")
print("After Adding 'Ali':", student_names)


student_names[2] = "Ahmed"
print("After Updating 'Fawwad' to 'Ahmed':", student_names)


print("All Students:", student_names)


student_names.remove("Saad") 
print("After Deleting 'Saad':", student_names)
```

```python
# Dictionary


student_records = {
    "Madni": 90,
    "Farukh": 85,
    "Fawwad": 88,
    "Saad": 92,
    "Fahad": 78
}


print("Initial Records:", student_records)


student_records["Ali"] = 95 
print("After Adding 'Ali':", student_records)


student_records["Fawwad"] = 90 
print("After Updating 'Fawwad' Grade:", student_records)


grade = student_records.get("Madni") 
print("Madni's Grade:", grade)


del student_records["Saad"]  
print("After Deleting 'Saad':", student_records)
```

```python
# Tuples

students = (("Madni", 80), ("Fawwad", 85), ("Farukh", 90))

print("Student Records:", students)

print()

print("Madni's grade is", students[0][1])  

print()

print( '''we cannot make changes in the tuples , 
if we want to update then we will have to create a new tuple to update or delete a record ''')

print()

students = (("Madni", 95), ("Fawwad", 100), ("Farukh", 82))

print("Updated Records:", students)

print()

students = (("Madni", 95), ("Farukh", 82))

print("Records after deletion:", students)

```

```python
# Sets

student_names = {'Madni', 'Farukh', 'Fawwad', 'Saad', 'Fahad'}


print("Student Names:", student_names)

print()
print('After removing a student name from the set')


student_names.remove("Saad")
print("Update No. 1: ", student_names)

print()
print('After adding a student name in the set')

student_names.add("Saeed")
print("Update No. 2: ", student_names)
```

## Strengths 

##### List = Ordered, mutable, allows duplicates
##### Tuples = Ordered, immutable, faster
##### Set = Unique elements, fast membership checks
##### Dictionary = Key-value pairs, fast lookups, flexible

## Weakness 

##### List = Slower for large datasets
##### Tuples = immutable
##### Set = Unordered, no duplicates allowed
##### Dictionary = Keys must be unique

```python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(6))  
```

```python
def log_function_call(func):
    def wrapper(n):
        result = func(n)
        print(f"Input: {n}, Output: {result}")
        return result
    return wrapper

@log_function_call
def fibonacci(n):
    if n <= 1:
        return n
    else:
        return fibonacci(n-1) + fibonacci(n-2)


fibonacci(6)
```

**How Recursion Works**  
Recursion is a programming technique where a function calls itself to solve smaller instances of the same problem. It typically involves two main components: a base case and a recursive step. The base case ensures that the recursion stops, while the recursive step breaks the problem into smaller subproblems. By repeatedly calling itself with modified inputs, the function eventually reaches the base case and returns the final result.

**The Benefit of Using Decorators for Logging**  
Decorators are a powerful feature in Python that allow you to modify or extend the behavior of functions without changing their actual code. When used for logging, decorators can automatically track and record details like function names, arguments, and return values. This approach keeps the logging logic separate from the core functionality, making the code cleaner and easier to maintain. Additionally, it enables consistent logging across multiple functions with minimal effort.

**Why Lambda Functions Can Be Useful in Such Contexts**  
Lambda functions are small, anonymous functions defined with minimal syntax. They are particularly useful in contexts like logging or recursion because they allow you to define simple operations inline without the need for a full function definition. For example, you can use a lambda function as part of a decorator to customize logging behavior dynamically. Their brevity and flexibility make them ideal for tasks where quick, lightweight functions are needed.

```python
class Book:
    def __init__(self, title, author, publication_year):
        """Initialize the attributes of the book."""
        self.title = title
        self.author = author
        self.publication_year = publication_year

    def __repr__(self):
        """Return a formatted string representation of the book."""
        return f"Book title = {self.title}, Author = {self.author}, Publication_year = {self.publication_year}"

    def update_year(self, new_year):
        """Update the publication year using a method."""
        self.publication_year = new_year
        print(f"Publication year updated to {self.publication_year}:")
        print('================================')

# Usage
book1 = Book("Urdu Book Class 9", "Iqbal Shakir", 1999)
print(book1)
print()
book1.update_year(2024)  # Updating the year
print(book1)
```
