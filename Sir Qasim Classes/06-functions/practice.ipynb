```python
def ned_student_card(id:int , name:str , fname:str, course:str, timing:str ) :
    card: str= f'''
    NED University
    Student Id = {id}
    Name = {name}
    Father Name = {fname}
    Course = {course}
    Timing = {timing}
    '''
    return card
```
