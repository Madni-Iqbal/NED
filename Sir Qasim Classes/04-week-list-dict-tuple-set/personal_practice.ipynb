```python
data1 = {
    "name": "Muhammad Qasim",
    "father_name": "Muhammad Aslam",
    "age": 30,
    "education": "MSDS",
    "city": "Karachi",
    "birth_city": "Karachi"
}

for key, value in data1.items():
    print(f"{key}: {value}")
```

```python
family = {
    "father": {
        "name": "Muhammad Aslam",
        "age": 60,
        "education": "MSDS",
        "skills": ["AI", "ML", "DL"]
    },
    "mother": {
        "name": "Naseem",
        "age": 55,
        "education": "MS"
    },
}

for role, details in family.items():
    print(f"{role.capitalize()}:")
    for key, value in details.items():
        # Handle lists differently
        if isinstance(value, list):
            value = ", ".join(value)  # Join list elements with commas
        print(f"  {key}: {value}")
    print()
```
